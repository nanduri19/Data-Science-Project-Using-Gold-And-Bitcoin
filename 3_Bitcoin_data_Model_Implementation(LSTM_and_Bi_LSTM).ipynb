```python
Coin_Price_Of_Historicaldata.replace([mu_Price_Of_Historicaldata.inf, -mu_Price_Of_Historicaldata.inf], mu_Price_Of_Historicaldata.nan, inplace=True)
Coin_Price_Of_Historicaldata.fillna(999, inplace=True)
```

```python
import tensorflow as txttflo
from keras.layers import Dense as txttdnse
from keras.layers import LSTM as txttlngtm
from sklearn.metrics import r2_score as txtr2scre
from sklearn.metrics import mean_squared_error as txtrmeaser
from sklearn.metrics import mean_absolute_error as txtrmeasqr
from sklearn.model_selection import train_test_split as txtrtsplt
from sklearn.preprocessing import MinMaxScaler as txtrtmnmxsler
from tensorflow.keras.layers import Bidirectional as price_of_histbidirct
```

```python
def His_SEGFn(txtIpXts_se_His, txtIpXts_n_His):
    vax_txtIpXts, vay_txtIpXts = [], []

    for RRr in range(len(txtIpXts_se_His)):
        Ad_txtIp = RRr + txtIpXts_n_His
        if Ad_txtIp > len(txtIpXts_se_His) - 1:
            break
        x_txtIpXts, y_txtIpXts = txtIpXts_se_His[RRr:Ad_txtIp], txtIpXts_se_His[Ad_txtIp]
        vax_txtIpXts.append(x_txtIpXts)
        vay_txtIpXts.append(y_txtIpXts)
    return mu_Price_Of_Historicaldata.array(vax_txtIpXts), mu_Price_Of_Historicaldata.array(vay_txtIpXts)

txtIpXts_n_His = 20

vax_txtIpXts_Tnn, vay_txtIpXts_Tnn = His_SEGFn(Coin_Price_Of_Historicaldata['BTC_Closing'][:-200].tolist(), txtIpXts_n_His)
vax_txtIpXts_Tss, vay_txtIpXts_Tss = His_SEGFn(Coin_Price_Of_Historicaldata['BTC_Closing'][-200:].tolist(), txtIpXts_n_His)
for total in range(len(vax_txtIpXts_Tss)):
    print(vax_txtIpXts_Tss[total], vay_txtIpXts_Tss[total])
```

```python
srtxtrtmnmxsler = txtrtmnmxsler()
vax_txtIpXts_Tnn_sca = srtxtrtmnmxsler.fit_transform(vax_txtIpXts_Tnn)
vax_txtIpXts_Tss_sca = srtxtrtmnmxsler.transform(vax_txtIpXts_Tss)

txtIpXtrrshped = vax_txtIpXts_Tnn_sca.reshape((vax_txtIpXts_Tnn_sca.shape[0], 1, vax_txtIpXts_Tnn_sca.shape[1]))
txtIpXtsrshped = vax_txtIpXts_Tss_sca.reshape((vax_txtIpXts_Tss_sca.shape[0], 1, vax_txtIpXts_Tss_sca.shape[1]))
```

 changing both Optimizer and activation function

LSTM

```python
txtoptmzers = ['adam', 'sgd', 'rmsprop', 'adadelta', 'adagrad', 'adamax', 'nadam']
txtoactvtns = ['relu', 'tanh', 'sigmoid']

for txtoptmzer in txtoptmzers:
  for txtoactvtn in txtoactvtns:
    txtmdle = txttflo.keras.models.Sequential()
    txtmdle.add(txttlngtm(32, activation=txtoactvtn, input_shape=(txtIpXtrrshped.shape[1], txtIpXtrrshped.shape[2])))
    txtmdle.add(txttdnse(1))
    txtmdle.compile(optimizer=txtoptmzer, loss='mse', metrics=['accuracy'])

    txtmdle.fit(txtIpXtrrshped, vay_txtIpXts_Tnn, epochs=30, batch_size=72, verbose=0)
    txtprdtns = txtmdle.predict(txtIpXtsrshped)
    txtr2sr = txtr2scre(vay_txtIpXts_Tss, txtprdtns)
    txtmsae = txtrmeaser(vay_txtIpXts_Tss, txtprdtns)
    txtmase = txtrmeasqr(vay_txtIpXts_Tss, txtprdtns)

    print(f'Optimizer is: {txtoptmzer}, activation is: {txtoactvtn}')
    print('R-squared:', txtr2sr)
    print('Mean Squared Error:', txtmsae)
    print('Mean Absolute Error:', txtmase)
    print('---')
```

Bidirectional LSTM

```python
txtoptmzers = ['adam', 'sgd', 'rmsprop', 'adadelta', 'adagrad', 'adamax', 'nadam']
txtoactvtns = ['relu', 'tanh', 'sigmoid']

for txtoptmzer in txtoptmzers:
  for txtoactvtn in txtoactvtns:
    txtmdle = txttflo.keras.models.Sequential()
    txtmdle.add(price_of_histbidirct(txttlngtm(32, activation=txtoactvtn, input_shape=(txtIpXtrrshped.shape[1], txtIpXtrrshped.shape[2]))))
    txtmdle.add(txttdnse(1))
    txtmdle.compile(optimizer=txtoptmzer, loss='mse')

    txtmdle.fit(txtIpXtrrshped, vay_txtIpXts_Tnn, epochs=30, batch_size=72, verbose=0)
    txtprdtns = txtmdle.predict(txtIpXtsrshped)
    txtr2sr = txtr2scre(vay_txtIpXts_Tss, txtprdtns)
    txtmsae = txtrmeaser(vay_txtIpXts_Tss, txtprdtns)
    txtmase = txtrmeasqr(vay_txtIpXts_Tss, txtprdtns)

    print(f'Optimizer is: {txtoptmzer}, activation is: {txtoactvtn}')
    print('R-squared:', txtr2sr)
    print('Mean Squared Error:', txtmsae)
    print('Mean Absolute Error:', txtmase)
    print('---')
```
